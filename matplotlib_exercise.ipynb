## 1問目
使うcsvはhist_data.csvです。下にコードを書いてください。

![問1](./imgs/matplotlib/plot_1.png)

```python
import matplotlib.pyplot as plt
import pandas as pd
import japanize_matplotlib


data = pd.read_csv('./data/hist_data.csv')


plt.hist(data['aデータ'], bins=20, alpha=0.7, label='aデータ', color='orange')
plt.hist(data['bデータ'], bins=20, alpha=0.5, label='bデータ', color='lightblue')

plt.xlabel('温度 (℃)')
plt.ylabel('度数')
plt.title('aデータとbデータの温度に関する度数分布')
plt.legend()

plt.show()
```

## 2問目
使うcsvはscatter_data.csvです。下にコードを書いてください。

![問2](./imgs/matplotlib/plot_2.png)

```python
import matplotlib.pyplot as plt
import pandas as pd
import japanize_matplotlib

data = pd.read_csv('./data/scatter_data.csv')


classes = {0: 'red', 1: 'blue', 2: 'green'}

for class_label, color in classes.items():
    subset = data[data['class'] == class_label]
    plt.scatter(subset['length (cm)'], subset['width (cm)'], label=f'class{class_label}', color=color)


plt.xlabel('length (cm)')
plt.ylabel('width (cm)')
plt.title('クラス 0～2 のlengthとwidthに関する散布図')
plt.legend()
plt.grid()

plt.show()
```

## 3問目
使うcsvはline_data.csvです。下に2つの画像を表示するコードを書いてください。

![問3_1](./imgs/matplotlib/plot_3_1.png)

![問3_2](./imgs/matplotlib/plot_3_2.png)

```python
import matplotlib.pyplot as plt
import pandas as pd
import japanize_matplotlib

# CSVデータの読み込み
data = pd.read_csv('./data/line_data.csv')

# 図のサイズを設定
fig, axs = plt.subplots(2, 1, figsize=(10, 8))

# 件数の推移（折れ線グラフ）
axs[0].plot(data['期間'], data['件数'], marker='o')
axs[0].set_title('期間ごとの件数の推移')
axs[0].set_xlabel('期間')
axs[0].set_ylabel('件数')
axs[0].grid(True)

# 費用の推移（棒グラフ）
axs[1].bar(data['期間'], data['費用'])
axs[1].set_title('期間ごとの費用の推移')
axs[1].set_xlabel('期間')
axs[1].set_ylabel('費用(円)')
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
